## Part 1: Compound-Interest Retirement Calculator

```python
import math


def round_cents(amount):
    """Round to the nearest cent (ordinary round-half-up bank rounding)."""
    return math.floor(amount * 100 + 0.5) / 100


principal = 87500.00
monthly_deposit = 125.00
monthly_rate = 0.0073
target_balance = 1_250_000.00

balance = principal
month = 0
total_extra_deposits = 0.0

while balance < target_balance:
    month += 1
    balance += monthly_deposit
    total_extra_deposits += monthly_deposit
    balance = round_cents(balance * (1 + monthly_rate))

    if month % 7 == 0:
        print(f"Month: {month:>3} | Balance: ${balance:>12,.2f}")

if month % 7 != 0:
    print(f"Month: {month:>3} | Balance: ${balance:>12,.2f}")

total_interest = balance - principal - total_extra_deposits

print(f"\nTotal number of months required: {month}")
print(f"Total interest earned: ${total_interest:,.2f}")
```

## Part 2: Conditional String Transformation

```python
def transform(s: str) -> str:
    words = s.split()

    # Rule 1: lowercase 'e' not first/last char of its word -> '3'
    step1_words = []
    for word in words:
        chars = list(word)
        n = len(chars)
        for i in range(n):
            if chars[i] == 'e' and i != 0 and i != n - 1:
                chars[i] = '3'
        step1_words.append(''.join(chars))

    # Rule 2: lowercase 'o' immediately followed by another vowel -> '0'
    vowels = "aeiou"
    step2_words = []
    for word in step1_words:
        chars = list(word)
        n = len(chars)
        for i in range(n):
            if chars[i] == 'o' and i + 1 < n and chars[i + 1] in vowels:
                chars[i] = '0'
        step2_words.append(''.join(chars))

    # Rule 3: remaining lowercase 'a' -> '@' on odd-numbered occurrences
    # (1st, 3rd, 5th, ...), counting left to right across the whole string
    step3_words = []
    a_count = 0
    for word in step2_words:
        chars = list(word)
        for i in range(len(chars)):
            if chars[i] == 'a':
                a_count += 1
                if a_count % 2 == 1:
                    chars[i] = '@'
        step3_words.append(''.join(chars))

    # Rule 4: reverse the order of the words
    reversed_words = step3_words[::-1]
    return ' '.join(reversed_words)


print(transform("the orange elephant ate an avocado"))
```

## Part 3: Gapminder Geo Dictionary

```python
import glob
import os
import pandas as pd

EXPECTED_FILENAME = 'Data Geographies - v1 - by Gapminder.xlsx'

if os.path.exists(EXPECTED_FILENAME):
    excel_path = EXPECTED_FILENAME
else:
    # Fall back to searching the current folder for any Gapminder xlsx,
    # in case it was saved under a slightly different name (e.g. a
    # browser-added "(1)" suffix) or different capitalization.
    matches = (
        glob.glob('*[Gg]apminder*.xlsx')
        + glob.glob('*[Gg]eographies*.xlsx')
    )
    if not matches:
        raise FileNotFoundError(
            f"Could not find '{EXPECTED_FILENAME}' (or anything similar) "
            f"in {os.getcwd()!r}. Make sure the .xlsx file is saved in "
            "the same folder as this notebook, then re-run this cell."
        )
    excel_path = matches[0]
    print(
        f"Note: expected filename not found, "
        f"using '{excel_path}' instead.\n"
    )

df = pd.read_excel(excel_path, sheet_name='List of countries etc')

print("Exact column names in the sheet:")
for col in df.columns:
    print(f"  {col!r}")

four_regions_values = sorted(df['four_regions'].dropna().unique().tolist())
income_col = 'World bank income group 2017'
income_group_values = df[income_col].unique().tolist()

print()
print("Unique values in 'four_regions':", four_regions_values)
print(f"Unique values in {income_col!r}:", income_group_values)

geo_dict = {}
for _, row in df.iterrows():
    geo_dict[row['geo']] = (
        row['four_regions'], row[income_col], row['name']
    )

print()
print("Number of entries in geo_dict:", len(geo_dict))
print("Example entry:", ('usa', geo_dict['usa']))
```

## Part 4: Americas Geo Codes

```python
americas_geo_codes = sorted(
    geo for geo, info in geo_dict.items() if info[0] == 'americas'
)

print("Geo codes in the Americas:", americas_geo_codes)
print("Number of entries:", len(americas_geo_codes))
```

## Part 5: High-Income Americas Countries

```python
# For loop version
high_income_americas_names = []
for geo, info in geo_dict.items():
    region, income_group, name = info
    if region == 'americas' and income_group == 'High income':
        high_income_americas_names.append(name)

# Equivalent list comprehension
high_income_americas_names_comp = [
    info[2] for geo, info in geo_dict.items()
    if info[0] == 'americas' and info[1] == 'High income'
]

assert (
    sorted(high_income_americas_names)
    == sorted(high_income_americas_names_comp)
)

high_income_americas_names = sorted(high_income_americas_names)
print("High-income Americas countries:", high_income_americas_names)

# One expression (no loop/list comprehension) for the longest name
longest_name = max(high_income_americas_names, key=len)
print(f"Longest name: {longest_name} ({len(longest_name)})")
```
